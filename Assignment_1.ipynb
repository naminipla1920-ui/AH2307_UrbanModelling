## Nested logit
Individuals are now assumed to make a joint choice of mode $m$ and destination $j$ conditional on their origin $i$ based on the utility of the combination.

<br>
Here, each destination $j$ represents a nest. Denoting $\lambda_j$ as the structural parameter for destination nest $j$, if $\epsilon^i_{j,m}$ is i.i.d. Gumbel distributed, the probability of optimal mode-destination pair ${j,m}$ is given by the MNL formula: <br>

$$P(j,m|i) = P(U^i_{j,m} \geq U^i_{j',m'}  \forall {j',m'}) =  \frac{e^{ V^i_{j,m}/ \lambda_j} (\sum\limits_{m \in M} e^{  V^i_{j,m}/ \lambda_j})^{\lambda_j-1}}{\sum\limits_{j' \in J} (\sum\limits_{m' \in M} e^{  V^i_{j',m'}/ \lambda_{j'}} )^{\lambda_{j'}}}$$ <br>

This probability can be decomposed as follows:
\begin{align}
P(j,m|i) = \frac{ e^{  V^i_{j,m}/ \lambda_j}} {\sum\limits_{m \in M} e^{  V^i_{j,m}/ \lambda_j}} \frac{(\sum\limits_{m \in M} e^{  V^i_{j,m}/ \lambda_j})^{\lambda_j}}{\sum\limits_{j' \in J} (\sum\limits_{m' \in M} e^{  V^i_{j',m'}/ \lambda_{j'}} )^{\lambda_{j'}}}
\end{align} <br>

The deterministic utility includes two parts $ V^i_{j,m}=V_{j}+V^i_{j,m}$, where $V_{j}$ depends only on variables that describe the destination and $V^i_{j,m}$ depends on variables that vary over alternatives within each destination.

\begin{align}
P(j,m|i) &= \frac{ e^{(V_{j}+V^i_{j,m})/ \lambda_j}} {\sum\limits_{m \in M} e^{(V_{j}+V^i_{j,m})/ \lambda_j}} \frac{(\sum \limits_{m \in M}e^{(V_{j}+V^i_{j,m})/ \lambda_j})^{\lambda_j}}{\sum\limits_{j' \in J} (\sum\limits_{m' \in M} e^{(V_{j'}+V^i_{j',m'})/ \lambda_{j'}})^{\lambda_{j'}}} \\ &=
\frac{e^{V_{j}/ \lambda_j} e^{V^i_{j,m}/ \lambda_j}}{e^{V_{j}/ \lambda_j} \sum\limits_{m \in M} e^{V^i_{j,m}/ \lambda_j}} \frac{e^{V_{j}} (\sum \limits_{m \in M} e^{ V^i_{j,m}/ \lambda_j} )^{\lambda_j}}{\sum\limits_{j' \in J} e^{V_{j'}} (\sum\limits_{m' \in M} e^{V^i_{j',m'}/ \lambda_{j'}})^{\lambda_{j'}}} \\ &= \frac{e^{V^i_{j,m}/ \lambda_j}}{\sum\limits_{m \in M} e^{V^i_{j,m}/ \lambda_j}} \frac{e^{V_{j}} e^{\lambda_j \ln \sum \limits_{m \in M} e^{V^i_{j,m}/ \lambda_j}}}{\sum\limits_{j' \in J} e^{V_{j'}} e^{\lambda_{j'} \ln \sum\limits_{m' \in M} e^{V^i_{j',m'}/ \lambda_{j'} }}}
\end{align} <br>

Let $I^i_j=\ln \sum \limits_{m \in M} e^{V^i_{j,m}/ \lambda_j}$ (the inclusive value or logsum), then $P(j,m|i)$ can be written as: <br>

$$P(j,m|i)= \frac{e^{V^i_{j,m}/ \lambda_j}}{\sum\limits_{m \in M} e^{V^i_{j,m}/ \lambda_j}} \frac{e^{V_j+\lambda_j I^i_j}}{\sum\limits_{j' \in J} e^{V_{j'}+\lambda_{j'} I^i_{j'}}}$$  <br>

The joint probability is now written as a product of the marginal and conditional probabilities:
$$P(j,m|i)= P(m|j,i)P(j|i) $$ <br>

# Expected utility
The expected utility (EU) is a measure of the total systematic utility provided by the choice set. For a given origin $i$, the expected utility across all destinations and modes is calculated as the log of the denominator of the upper-level choice:

$$EU^i = \ln \left( \sum_{j \in J} e^{V_{j} + \lambda_{j} I^i_{j}} \right)$$

This $EU$ can be used as a welfare measure when analyzing the change between two different scenarios. If cost enters the utility functions, the difference in expected utility between scenario 0 and scenario 1 can be translated into monetary terms to measure the Consumer Surplus (CS):<br>

$$ CS = \frac{EU^0 - EU^1}{\gamma_{\text{cost}}} $$
*(Note: Since $\gamma_{\text{car}} = \gamma_{\text{pt}} = -0.1$, the marginal utility of cost $\gamma_{\text{cost}}$ is constant across modes).*
